```python
import pandas as pd
from matplotlib import pyplot as plt
import seaborn as sns

shipments_df= pd.read_csv('ApplianceShipments.csv')

#Question 1, (1)

quarter_df = shipments_df[shipments_df['Year'] == 1987]
sns.lineplot(data=quarter_df, x='Quarter', y='Shipments', marker='o')
plt.title('1987 Quarterly Appliance Shipments')
plt.xlabel('Quarter')
plt.ylabel('Shipments(in millions of dollars)')
plt.xticks([1, 2, 3, 4])
plt.show()
```

```python
#Question 1, (2)

sns.lineplot(data=shipments_df, x='Year', y='Shipments', marker='o', hue='Quarter')
plt.title('Quarterly Appliance Shipments')
plt.xlabel('Year')
plt.ylabel('Shipments(in millions of dollars)')
plt.xticks(shipments_df['Year'])
plt.show()

# Q2 shows consistent increase in shipment across the years followed by Q3.
# Q4 and Q1 show lowest shipment quarter of the year.
# Across all quarters , shipment shows a upward trend from 1985 1987, and Q2 and Q4 continued to increase in 1989.
# Q1 experienced a sharp decline compared to its peak in 1988.
```

```python
#Question 1, (3)

yearly_shipments = shipments_df.groupby("Year", as_index=False)["Shipments"].sum()

sns.lineplot(data=yearly_shipments, x='Year', y='Shipments', marker='o')
plt.title('Total Yearly Appliance Shipments')
plt.xlabel('Year')
plt.ylabel('Shipments(in millions of dollars)')
plt.xticks(shipments_df['Year'])
plt.show()

#The total shipments seems to be increasing from year 1985, however there's a drop in 1988 and again spikes up in 1989.
```

```python
#Question 2, 1- Summary statistics(1)

airbnb_df= pd.read_csv('AB_NYC_2019.csv')

brooklyn_df = airbnb_df[airbnb_df['neighbourhood_group'] == 'Brooklyn']
# brooklyn_df
print("Number of Airbnb listings in Brooklyn: ", len(brooklyn_df))
```

```python
#Question 2, 1- Summary statistics (2)

longer_than_30_nights = airbnb_df[airbnb_df['minimum_nights'] > 30]
# longer_than_30_nights
print("Number of Airbnb listings that require stays longer than 30 nights: ", len(longer_than_30_nights))
```

```python
#Question 2, 1- Summary statistics (3)

average_price_Manhattan = airbnb_df[airbnb_df['neighbourhood_group']=='Manhattan']['price'].mean()
print("Average price of Manhattan Airbnb listings: ", average_price_Manhattan)
```

```python
#Question 2, (2)

sns.scatterplot(data=airbnb_df, x='number_of_reviews', y='price', alpha=0.3)
plt.title('Number of Reviews vs. Price')
plt.xlabel('Number of Reviews')
plt.ylabel('Price')
plt.show()

# Most high priced listings(above $800) have zero or few reviews, whereas price ranging between $0-$300/night has dense concentration of points.
# Expensive listings are reviwed significantly less compared to lower or moderately priced listings.
# High end listings are booked less frequently because of their high cost, resulting in fewer total guest stays.
# Or the extreme high prices might also be set by new hosts to experiment market rates , which hasn't accumulated any reviews yet.
```

```python
#Question 2, (3)

minimum_nights= airbnb_df[airbnb_df['minimum_nights'] < 30]

sns.scatterplot(data=minimum_nights, x='minimum_nights', y='price', alpha=0.3, hue='room_type')
plt.title('Minimum Nights vs. Price')
plt.xlabel('Minimum nights')
plt.ylabel('Price')
plt.legend(title='Room Type')
plt.show()

# Most listings cluster around shorter night stays(1-7 nights). Entire home/apt listings show higher price across all minimum stays, whereas private/shared room have lower prices regardless of stay length.
# Hosts offering short minimum stays can charge premium price due to increased demand.
# Hosts can also offer slight discounts for longer minimum stays(eg., 7-14nights) to attract guests and get profit.
```

```python
#Question 2, (4)

price_df = airbnb_df[airbnb_df['price'] < 1500]

sns.boxplot(data=price_df, x='neighbourhood_group', y='price')
plt.title('Price Distribution by Neighbourhood Group(Price < $1500)')
plt.xlabel('Neighbourhood Group')
plt.ylabel('Price $(USD)')
plt.show()

# Manhattan has highest median price among all neighbourhood groups, followed by Brooklyn.
# Manhattan appears to have widest range of prices extending upto $1500.
# Hosts in Manhattan can charge higher prices because guests seem to pay more, whereas hosts in Queens or Bronx should keep their prices low to attract budget conscious people.
# Platform Managers can market Manhattan as luxury location to wealthy travellers, whereas for Queens and Bornx managers should promote cheap deals.
```

```python
#Question 2, (5)

sns.violinplot(data=airbnb_df, x='room_type', y='price')
plt.title('Price Distribution by Room Type')
plt.xlabel('Room Type')
plt.ylabel('Price')
plt.show()

# Entire home/apt has highest median price than the private and shared room.
# Entire home/apt has widest and borades range of pricings(from cheap to expensive).
# Guests who want to save money should look out for private or shared rooms.
# Hosts who rent out Entire home/apt can charge extra if they offer some extra nice features.
# The platform manager can suggest optimal prices for new hosts based on their room type.
```

```python
#Question 2, (6)

sns.barplot(data=airbnb_df[airbnb_df['price'] < 1500], x='neighbourhood_group', y='price', hue='room_type',)
plt.title('Average Listing Price by Neighbourhood Group and Room Type')
plt.xlabel('Neighbourhood Group')
plt.ylabel('Price ($ USD)')
plt.legend(title='Room Type')
plt.show()

# This bar chart shows how listing prices vary across neighbourhood group and room type.
# Across all five neighbourhood groups, Entire home/apt commands highest price with Manhattan leading over $200/night on average.
# Private rooms offer mid tier option ($70-$120)/night regradless of neighbourhood locations.
# Platform managers can advertise private rooms in expensive areas like Manhattan so budget travellers can still afford to visit.
```
