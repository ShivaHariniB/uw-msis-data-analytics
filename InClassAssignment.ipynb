```python
import pandas as pd
housing_df = pd.read_csv("housing.csv")
```

```python
#1
print(housing_df['ocean_proximity'])
```

```python
#2
import matplotlib.pyplot as plt
import seaborn as sns

sns.scatterplot(data=housing_df, x='housing_median_age', y= 'median_house_value')
plt.title('Relationship between Housing Age and House Values')
plt.xlabel('Housing Median Age')
plt.ylabel('House Median Values')
plt.show()
```

#2 Insights:
shows prices may vary regardless of old or new homes, however older homes show skight increase in price cap, becuase old homes are located in prime , well establshed or histrpcial locations

```python
#3

sns.scatterplot(data=housing_df, x='longitude', y= 'latitude', hue='median_house_value', alpha=0.7)
plt.title('Location of Houses')
plt.xlabel('Longitude')
plt.ylabel('Latitude')
plt.show()
```

```python
#4

sns.scatterplot(data=housing_df, x='longitude', y= 'latitude', hue='ocean_proximity', alpha=0.5)
plt.title('Location of Houses')
plt.xlabel('Longitude')
plt.ylabel('Latitude')
plt.show()
```

#4 Insights:
 Inland homes are spread out across interior with much lower prices, homes in ocean areas are tightly clustered in high value areas. this proves that ocean proximity and high neighbourhood prices are factors driving up home prices.

```python
#5

sns.boxplot(data=housing_df, x='ocean_proximity', y='median_house_value')
plt.title('Relationship between Ocean Proximity and House Values')
plt.xlabel('Ocean Proximity')
plt.ylabel('House Median Values')
plt.show()
```

#5 Insights:
Island has the highest typical house value. the median line for island is located around $410,000 which is higher than every other category.

Near bay shows greatest variation in house values, the middle 50% spread of data is shown by Q3-Q1= IQR. Near bay has highest IQR of $180,000.
